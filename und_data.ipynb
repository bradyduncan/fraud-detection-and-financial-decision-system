```python
import pandas as pd

df = pd.read_csv('data/processed/predictions.csv')

# Basic info
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst 5 rows:")
print(df.head())

# Demo user breakdown
print("\n--- Demo User Transaction Counts ---")
print(df[df['demo_user'] != 'Unknown'].groupby(['demo_user', 'demo_card']).agg(
    transactions=('fraud_probability', 'count'),
    fraud_flagged=('model_decision', 'sum'),
    actual_fraud=('isFraud', 'sum'),
    avg_fraud_prob=('fraud_probability', 'mean')
).round(4))

# Overall stats
print("\n--- Overall Prediction Stats ---")
print(f"Total transactions     : {len(df):,}")
print(f"Flagged as fraud       : {df['model_decision'].sum():,}")
print(f"Actual fraud           : {df['isFraud'].sum():,}")
print(f"Unknown (non-demo) rows: {(df['demo_user'] == 'Unknown').sum():,}")

# Date range
print(f"\nDate range: {df['transaction_date'].min()} → {df['transaction_date'].max()}")

# Missing values
print("\n--- Missing Values ---")
print(df.isnull().sum()[df.isnull().sum() > 0])
```
